```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv(r"..\data\processed\tickets.csv")
```

```python
df.head()
```

```python
df.columns
```

#### Category Analysis

```python
df["category"].value_counts(normalize=True).mul(100).round(2).astype(str) + "%"
```

```python
pd.crosstab(df["category"], df["priority"], 
            normalize="index").mul(100).round(2).astype(str) + "%" 
```

```python
pd.crosstab(df["category"], df["language"],
            normalize="index").mul(100).round(2).astype(str) + "%"
```

```python
table = pd.crosstab(df["category"], df["priority"])

chi2, p_value, dof, expected = chi2_contingency(table)

print(f"Chi-square: {chi2:.2f}")
print("p-value: ", p_value)
print("Degrees of freedom: ", dof)
```

```python
if p_value < 0.05:
    print("There is a statistically significant relationship between category and priority.")
else:
    print("There is insufficient statistical evidence for a significant association.")
```

```python
n = table.to_numpy().sum()
r, k = table.shape

cramers_v = np.sqrt(
    (chi2 / n) / min(r - 1, k - 1)
)

print(f"Cramer's V: {cramers_v:.2f}")
```

The Chi-square test showed a statistically significant association between ticket category and priority, as the p-value was far below 0.05. The distribution of priority levels is not the same across all categories; for example, Service Outages and Maintenance and Technical Support tickets have higher proportions of high-priority cases. Cramér’s V was 0.29, indicating a moderate association between category and priority.

The chi-square is not reliable for this relation.

```python
if p_value < 0.05:
    print("There is a statistically significant relationship between category and priority.")
else:
    print("There is insufficient statistical evidence for a significant association.")
```

```python
n = table.to_numpy().sum()
r, k = table.shape

cramers_v = np.sqrt(
    (chi2 / n) / min(r - 1, k - 1)
)

print(f"Cramer's V: {cramers_v:.2f}")
```

The Chi-square test and Cramér’s V are not appropriate for examining the relationship between `category` and `response` in this dataset. The `response` column contains free-text values with a very high number of unique responses, resulting in a large and sparse contingency table. Therefore, the non-significant p-value and Cramér’s V of 1.00 should not be interpreted as evidence of either no relationship or a perfect relationship. Instead, responses should first be converted into meaningful groups, such as topics, intents, or sentiment categories, before performing the analysis.
